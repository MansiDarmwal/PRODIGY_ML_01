## House Price Prediction using Linear Regression

For this task, I built a simple model to predict house prices using the living area, number of bedrooms and number of bathrooms.

I kept the approach straightforward and followed the usual machine learning workflow from loading the data to checking the final predictions.

## Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## Loading the data

First, I loaded the CSV file and looked at a few rows to make sure everything was read correctly.

```python
df = pd.read_csv("PRODIGY_ML_01_dataset.csv")

df.head()
```

## Understanding the dataset

Before building the model, I checked the number of rows, the columns and whether there were any missing values.

```python
print("Dataset size:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

## Basic statistics

This gives a quick idea of the values present in the dataset.

```python
df.describe().round(2)
```

## Selecting the features

The task asks us to use:

- `GrLivArea` — living area in square feet
- `BedroomAbvGr` — number of bedrooms
- `FullBath` — number of full bathrooms

The value we want to predict is `SalePrice`.

```python
X = df[["GrLivArea", "BedroomAbvGr", "FullBath"]]
y = df["SalePrice"]

print("Selected features:", list(X.columns))
print("Target: SalePrice")
```

## Splitting the data

I used 80% of the data for training and 20% for testing. The test data is kept separate so that the model can be checked on data it has not seen before.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## Training the model

Now I trained a Linear Regression model using the training data.

```python
model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Model trained successfully.")
```

## Looking at the learned coefficients

The coefficients show the relationship learned by the model for each of the three features.

```python
print(f"Intercept: ${model.intercept_:,.2f}")

for feature, coefficient in zip(X.columns, model.coef_):
    print(f"{feature}: ${coefficient:,.2f}")
```

## Comparing actual and predicted prices

Here I compared some of the model's predictions with the actual prices from the test set.

```python
comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred.round(2)
})

comparison.head(10)
```

## Model evaluation

I used four metrics to evaluate the regression model:

- **MAE** — average absolute prediction error
- **MSE** — squared prediction error
- **RMSE** — error in the same unit as the house price
- **R²** — how much variation in the target is explained by the model

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE : ${mae:,.2f}")
print(f"MSE : ${mse:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"R²  : {r2:.4f}")
```

## Actual vs predicted prices

The dashed line represents a perfect prediction. Points closer to the line mean the predicted price is closer to the actual price.

```python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, alpha=0.7)

low = min(y_test.min(), y_pred.min())
high = max(y_test.max(), y_pred.max())
plt.plot([low, high], [low, high], linestyle="--")

plt.xlabel("Actual Sale Price")
plt.ylabel("Predicted Sale Price")
plt.title("Actual vs Predicted House Prices")
plt.tight_layout()
plt.show()
```

## Final thoughts

The model gives a reasonable baseline for predicting house prices from the three selected features.

The main thing I noticed is that house price depends on much more than just size, bedrooms and bathrooms. A more complete model could include information such as location, overall condition, year built and garage space.
